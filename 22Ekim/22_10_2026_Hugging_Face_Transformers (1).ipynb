```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

Model_adı="Qwen/Qwen3.5-0.8B"
cihaz="cuda" if torch.cuda.is_available() else "cpu"
tokenizer=AutoTokenizer.from_pretrained(Model_adı, trust_remote_code=True)
print(f"kullanılan cihaz: {cihaz}")
model=AutoModelForCausalLM.from_pretrained(Model_adı,trust_remote_code=True).to(cihaz)
model.eval()
print(f"parametre sayısı: {model.num_parameters()}")
```

```python
ornek="Yapay zeka dünyayı değişiyor."
tokenlar=tokenizer.tokenize(ornek)
idler=tokenizer.encode(ornek)
print(f"tokenler: {tokenlar}")
print(f"idler: {idler}")
print(f"geri çözülmüş:",tokenizer.decode(idler))
```

```python
def sor(soru, sistem="Sen yardımsever bir asistansın. Türkçe ve kısa cevap ver.", max_yeni_token=200, temperature=0.7, top_p=0.9):
    mesajlar = [
        {"role": "system", "content": sistem},
        {"role": "user", "content": soru}
    ]
    metin = tokenizer.apply_chat_template(mesajlar, tokenize=False, add_generation_prompt=True)
    tokenlar = tokenizer(metin, return_tensors="pt").to(cihaz)
    with torch.no_grad():
        cikti = model.generate(
            **tokenlar,
            max_new_tokens=max_yeni_token,
            do_sample=temperature > 0,
            temperature=temperature if temperature > 0 else None,
            top_p=top_p if temperature > 0 else None,
            pad_token_id=tokenizer.eos_token_id
        )
    yeni = cikti[0][tokenlar["input_ids"].shape[1]:]
    return tokenizer.decode(yeni, skip_special_tokens=True)

print(sor("sevgiliye her hafta çiçek alınır mı ?"))
```

```python
sorular=[
    "Fatih 3'ü bir ada sever mi?",
    "Komşular iyi midir",
    "Maria mı haklıydı yoksa yoksa Ragıp Efendi mi ",
    "Evlenmeden önce kaç yıl sevgili kalınmalı"
]

for s in sorular:
  print(f"❓ {s}\n ❤️‍🩹 sor{sor(s)}\n"+"--"*4)
```

```python
soru="Aşk ve kasımpatı ile ilgili bir şiir yaz"
for t in [0,0.7,1.3]:
  print(f"temperature={t}:{sor(soru,temperature=t,max_yeni_token=60)}\n")
```

### Daha Güçlü Bir Model Yükleme (Qwen-2.5-7B-Instruct)

Şu anki 0.8B parametreli model yerine, donanımınızın gücünü tam kullanacak **7 Milyar parametreli** çok daha zeki bir modele geçiş yapalım:

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# 7 Milyar parametreli çok daha güçlü model
yeni_model_adi = "Qwen/Qwen2.5-7B-Instruct"
cihaz = "cuda" if torch.cuda.is_available() else "cpu"

print(f"{yeni_model_adi} yükleniyor, lütfen bekleyin...")

tokenizer = AutoTokenizer.from_pretrained(yeni_model_adi, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    yeni_model_adi,
    torch_dtype=torch.float16, # Bellek optimizasyonu için FP16 kullanıyoruz
    device_map="auto"
)

print(f"Yeni model başarıyla yüklendi! Cihaz: {cihaz}")
print(f"Parametre sayısı: {model.num_parameters():,}")
```

```python
# Yeni güçlü modelimizle bir deneme yapalım
print(sor("Yapay zeka modellerinde parametre boyutunun (örneğin 0.8B ile 7B arasındaki farkın) üretime etkisi nedir?"))
```

```python
# sor fonksiyonunun yeni yüklenen 7B modeli kullanması için değişkenleri eşitleyelim
model = model
tokenizer = tokenizer

# Şimdi yeni modelle testlerimizi yapalım
print("--- Yeni 7B Model Testi ---")
print("Soru: Sevgiliye her hafta çiçek alınır mı?\n")
print(sor("sevgiliye her hafta çiçek alınır mı ?"))
print("\n" + "="*40 + "\n")
print("Soru: Maria mı haklıydı yoksa Ragıp Efendi mi?\n")
print(sor("Maria mı haklıydı yoksa yoksa Ragıp Efendi mi "))
```

```python
class Sohbet:
  def __init__(self,sistem="Sen yardımsever bir asistan. Türkçe cevap ver"):
    self.gecmis=[{"role":"system","content":sistem}]

  def mesaj(self,kullanici_mesaji,max_yeni_token=200):
    self.gecmis.append({"role":"user","content":kullanici_mesaji})
    metin=tokenizer.apply_chat_template(self.gecmis,tokenize=False,add_generation_prompt=True)
    tokenlar=tokenizer(metin,return_tensors="pt").to(cihaz)
    with torch.no_grad():
      cikti=model.generate(**tokenlar,max_new_tokens=max_yeni_token,do_sample=True,temperature=0.2,top_p=0.9,pad_token_id=tokenizer.eos_token_id)
    cevap=tokenizer.decode(cikti[0][tokenlar["input_ids"].shape[1]:],skip_special_tokens=True)
    self.gecmis.append({"role":"assistant","content":cevap})
    return cevap

sohbet=Sohbet()
print("🤖",sohbet.mesaj("benim ismim eren altun "))
```
