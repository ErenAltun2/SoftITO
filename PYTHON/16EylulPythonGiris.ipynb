```python
def toplam_hesapla(*sayilar):   # * args ıstedıgımız kadar arguman gondermemızı saglar
  toplam=0
  for sayi in sayilar:
    toplam+=sayi
  return toplam

print(toplam_hesapla(1,2,3,4,5))

def profil_olustur(**bilgiler):   # ** fonksıyona anahtar deger atamayı saglar
  for anahtar,deger in bilgiler.items():
    print(f"{anahtar}: {deger}")

profil_olustur(ad="Eren",soyad="Altun",yas=22)
```

```python
# ARABA SINIFI
"""
SORU-1
marka, model ve hiz (km/s) bilgilerini tutan bir Araba sınıfı oluşturun.
hızlandır(artis) metodu ile hızı artırın.
bilgi_goster() metodu ile araç bilgilerini yazdırın.
"""
class araba:
  def __init__ (self,marka,model,hiz):
    self.marka=marka
    self.model=model
    self.hiz=hiz

  def hizlandir(self):
    return self.hiz+5

  def bilgi_goster(self):
    return print(self.marka,self.model,self.hiz)

araba1=araba("Hyundai","i20",110)
araba1.bilgi_goster()
araba1.hizlandir()
```

```python
"""
SORU-2
uzunluk ve genislik bilgilerini tutan bir Dikdortgen sınıfı oluşturun.
alan_hesapla() metodu ile alanı hesaplayın (uzunluk * genislik).
cevre_hesapla() metodu ile çevreyi hesaplayın (2 * (uzunluk + genislik)).
"""
class dikdortgen:
  def __init__(self,uzunluk,genislik):
    self.uzunluk=uzunluk
    self.genislik=genislik

  def alan_hesapla(self):
    return self.uzunluk*self.genislik

  def cevre_hesapla(self):
    return (self.uzunluk+self.genislik)*2

dikdortgen1=dikdortgen(10,5)
print(dikdortgen1.alan_hesapla())
print(dikdortgen1.cevre_hesapla())
```

```python
"""
SORU-3
ad, yazar ve sayfa_sayisi bilgilerini tutan bir Kitap sınıfı oluşturun.
ozet() metodu ile "Suç ve Ceza - Dostoyevski (671 sayfa)" formatında bilgi döndürün.
"""

class kitap:
  def __init__(self,ad,yazar,sayfa_sayisi):
     self.ad=ad
     self.yazar=yazar
     self.sayfa_sayisi=sayfa_sayisi

  def ozet(self):
    return print(self.ad,self.yazar,f"({self.sayfa_sayisi} sayfa)")

kitap1=kitap("Yaban","Yakup Kadri Karaosmanoğlu",214)
kitap1.ozet()
```

```python
class Urun:
  def __init__(self,isim,fiyat,stok_adedi):
    self.isim=isim
    self.fiyat=fiyat
    self.stok_adedi=stok_adedi

  def toplam_hesapla(self):
    return self.fiyat*self.stok_adedi

  def stok_durum(self):
    if self.stok_adedi==0:
      return "Tükendi"
    elif self.stok_adedi <10:
      return "Stok Azalmış"
    else:
      return "yeterlı stok"

urun1=Urun("elma",23,10)
print(urun1.toplam_hesapla())
print(urun1.stok_durum())
```
