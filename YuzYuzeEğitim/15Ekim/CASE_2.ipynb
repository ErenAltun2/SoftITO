```python
sentences = [
    # Teknoloji ve Bilim
    ["yapay", "zeka", "ve", "derin", "ogrenme", "teknoloji", "dunyasini", "degistiriyor"],
    ["yazilim", "gelistiricileri", "python", "ve", "algoritma", "konularinda", "uzmanlasir"],
    ["bilgisayar", "ve", "internet", "bilgiye", "erisimde", "buyuk", "kolaylik", "saglar"],
    ["veri", "bilimi", "makine", "ogrenmesi", "ve", "istatistik", "temellerine", "dayanir"],
    ["robotik", "ve", "uzay", "arastirmalari", "gelecegin", "teknolojilerini", "bicimlendirir"],

    # Doğa, Coğrafya ve Şehirler
    ["turkiye", "ve", "fransa", "avrupa", "kitasinda", "yer", "alan", "ulkelerdir"],
    ["ankara", "turkiye", "ulkesinin", "baskentidir"],
    ["paris", "fransa", "ulkesinin", "baskentidir"],
    ["istanbul", "ve", "izmir", "turkiyenin", "buyuk", "sehirlerindendir"],
    ["deniz", "orman", "dag", "ve", "nehir", "doga", "guzellikleridir"],
    ["gunes", "ve", "yildizlar", "uzayda", "bulunan", "gok", "cisimleridir"],

    # Yiyecek, İçecek ve Günlük Yaşam
    ["sabahlari", "kahve", "veya", "cay", "icmek", "beni", "canlandirir"],
    ["kahve", "ve", "cay", "sikca", "tuketilen", "sicak", "iceceklerdir"],
    ["ekmek", "ve", "su", "temel", "besin", "ihtiyaclarimiz", "arasindadir"],
    ["elma", "ve", "muz", "saglikli", "meyvelerdir"],

    # Meslekler ve İlişkiler
    ["doktor", "hastanede", "hastalari", "tedavi", "eder"],
    ["ogretmen", "okulda", "ogrencilere", "ders", "anlatir"],
    ["muhendis", "yeni", "projeler", "ve", "tasarimlar", "gelistirir"],
    ["kral", "erkek", "yoneticidir"],
    ["kralice", "kadin", "yoneticidir"],
    ["anne", "baba", "ve", "cocuk", "aile", "bireyleridir"],
]

print("Cümle sayısı:", len(sentences))
```

```python
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    sg=1,          # 1 = skip-gram, 0 = CBOW (hocanın kodunda 0'dı)
    epochs=200,
    seed=42,
    workers=1      # sonuçların her çalıştırmada aynı çıkması için
)

print("Sözlük boyutu:", len(model.wv))
```

```python
print(model.wv["kral"][:5])
print("Vektör boyutu:", model.wv["kral"].shape)
```

```python
for kelime in ["kral", "kahve", "turkiye", "ogretmen"]:
    print(kelime, "->", model.wv.most_similar(kelime, topn=3))
```

```python
ciftler = [
    ("kral", "kralice"),
    ("kahve", "cay"),
    ("ankara", "paris"),
    ("elma", "muz"),
    ("kral", "muz"),
]
for a, b in ciftler:
    print(f"{a} - {b}: {model.wv.similarity(a, b):.4f}")
```

```python
print(model.wv.most_similar(positive=["kral", "kadin"], negative=["erkek"], topn=3))
```
