```python
!pip install gensim
from gensim.models import Word2Vec

cümleler =[
    ["kedi","süt","içer"],
    ["köpek","et","yer"],
    ["kedi","fare","kovalar"],
    ["köpek", "kedi", "kovalar"],
    ["inek","süt","verir"],
    ["arı","bal","yapar"]
]
```

```python
#model eğitelim
model = Word2Vec(
    sentences=cümleler,
    vector_size=30,
    window=2,
    min_count=1,
    sg=0,
    epochs = 100
)
```

```python
print(model.wv["kedi"][:5])
```

```python
print(model.wv.most_similar("kedi",topn=3))
```

```python
print(model.wv.similarity("kedi","fare"))
```

```python
model.save("word2vec.model")
model = Word2Vec.load("word2vec.model")
```

# glove

```python
import gensim.downloader as api


model = api.load("glove-wiki-gigaword-50")


print(model.most_similar("king"))


print(model.similarity("cat", "dog"))

print(model.most_similar(positive=["king", "woman"], negative=["man"], topn=1))
```
