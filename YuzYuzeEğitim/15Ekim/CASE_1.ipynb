```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

cumleler = [
    "İstanbul'da sabahları trafik çok yoğun oluyor.",
    "Yoğun trafik yüzünden birçok kişi işe metroyla gidiyor.",
    "Metro, İstanbul'da en hızlı ulaşım seçeneklerinden biri.",
    "Vapur ise boğazı geçmek için keyifli bir ulaşım yolu.",
    "Sabahları vapurda çay içmek İstanbullular için bir alışkanlık.",
    "Hafta sonları trafik azalıyor ve şehir daha sakin oluyor.",
    "Sakin bir hafta sonunda boğaz kıyısında yürüyüş yapmak çok güzel.",
]

# Türkçe için küçük harfe çevirme (İ -> i, I -> ı)
def turkce_kucult(metin):
    return metin.replace("İ", "i").replace("I", "ı").lower()

vectorizer = TfidfVectorizer(preprocessor=turkce_kucult)
X = vectorizer.fit_transform(cumleler)

# Sonuçları tablo olarak görmek için
df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
print(df.round(2))

# Her cümle için en yüksek skorlu 3 kelime
kelimeler = vectorizer.get_feature_names_out()
for i, satir in enumerate(X.toarray()):
    en_iyi = satir.argsort()[-3:][::-1]
    print(f"Cümle {i+1}:", [(kelimeler[j], round(satir[j], 2)) for j in en_iyi])
```
